**Note:** `recid.csv` has no missing values, so NaNs are randomly introduced (5%) into a few columns to carry out Q8–Q18. `black` is treated as the categorical column.

```python
rng = np.random.default_rng(42)
df_na = df.copy()
for col in ["educ", "age", "priors", "black", "tserved"]:
    df_na.loc[rng.random(len(df_na)) < 0.05, col] = np.nan
```

**Q8 & Q10. Number of missing values in each column**

```python
df_na.isna().sum()
```

**Q9. Total number of missing values**

```python
print("Total missing:", df_na.isna().sum().sum())
```

**Q11. Rows containing at least one missing value**

```python
rows_na = df_na[df_na.isna().any(axis=1)]
print("Rows with missing values:", len(rows_na)); rows_na.head(10)
```

**Q12. Remove rows containing missing values**

```python
df_dropna = df_na.dropna()
print("Before:", df_na.shape, "| After:", df_dropna.shape); df_dropna.head(10)
```

**Q13. Mean imputation – `educ`**   **Q14. Median imputation – `age`**   **Q15. Mode imputation – `black` (categorical)**

```python
df_fill = df_na.copy()
df_fill["educ"]  = df_fill["educ"].fillna(df_fill["educ"].mean())
df_fill["age"]   = df_fill["age"].fillna(df_fill["age"].median())
df_fill["black"] = df_fill["black"].fillna(df_fill["black"].mode()[0])
print(df_fill[["educ", "age", "black"]].isna().sum())
```

**Q16. Forward fill (like zoo::locf)**

```python
df_ffill = df_na.ffill()
print("Remaining NaN after forward fill:\n", df_ffill.isna().sum()[lambda s: s > 0])
```

**Q17. Backward fill**

```python
df_bfill = df_na.bfill()
print("Remaining NaN after backward fill:\n", df_bfill.isna().sum()[lambda s: s > 0])
```

**Q18. Replace remaining missing values with a constant (0)**

```python
df_const = df_ffill.fillna(0)          # leading NaNs left by forward fill -> 0
print("Total NaN left:", df_const.isna().sum().sum())
```

**Q19. Remove duplicate records** (ignoring the `rownames` id column)

```python
cols = [c for c in df.columns if c != "rownames"]
print("Duplicates:", df.duplicated(subset=cols).sum())
df_nodup = df.drop_duplicates(subset=cols)
print("Before:", df.shape, "| After:", df_nodup.shape)
```

**Q20. Convert appropriate variables into factors (category)**

```python
df2 = df.copy()
for c in ["black","alcohol","drugs","super","married","felon","workprg","property","person","cens"]:
    df2[c] = df2[c].astype("category")
df2.dtypes
```

**Q21. Rename column names**

```python
df2 = df2.rename(columns={"super": "supervised", "workprg": "work_program", "tserved": "months_served"})
print(list(df2.columns))
```

**Q22. New derived column** (age in years)

```python
df2["age_years"] = (df2["age"] / 12).round(1)
df2[["age", "age_years"]].head()
```

**Q23. Summary statistics (summary)**

```python
df2.describe(include="all").T
```

**Q24–Q28. Min, Max, Range, Variance, Standard Deviation, IQR** (numerical variable: `months_served`; then for all numeric columns)

```python
x = df2["months_served"]
print("Minimum :", x.min()); print("Maximum :", x.max()); print("Range   :", x.max() - x.min())
print("Variance:", round(x.var(), 3)); print("Std Dev :", round(x.std(), 3))
print("IQR     :", x.quantile(0.75) - x.quantile(0.25))

num = df2[["priors","educ","rules","age","months_served","follow","durat"]]
pd.DataFrame({"Min": num.min(), "Max": num.max(), "Range": num.max()-num.min(), "Variance": num.var(),
              "StdDev": num.std(), "IQR": num.quantile(.75)-num.quantile(.25)}).round(3)
```
