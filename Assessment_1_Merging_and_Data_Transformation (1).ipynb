**Dataframe Operations** – create `dataframe1`, `dataframe2`, `dataframe3` (10 records each).

```python
import pandas as pd, numpy as np

dataframe1 = pd.DataFrame({
    "ProductID":  [101,102,103,104,105,106,107,108,109,110],
    "NameofProd": ["Laptop","Mouse","Keyboard","Monitor","Printer","Scanner","Tablet","Phone","Camera","Speaker"],
    "Price":      [55000,500,1200,9500,7500,6200,18000,22000,30000,3500]})

dataframe2 = pd.DataFrame({
    "ProductID": [105,106,107,108,109,110,111,112,113,114],
    "Company_Location": ["Chennai","Mumbai","Delhi","Bengaluru","Pune","Hyderabad","Kolkata","Jaipur","Kochi","Noida"]})

dataframe3 = pd.DataFrame({
    "ProductID":  [111,112,113,114,115,116,117,118,119,120],
    "NameofProd": ["Router","Webcam","Headset","Pendrive","HardDisk","SSD","Projector","Charger","PowerBank","SmartWatch"],
    "Price":      [2500,1800,2200,600,4500,5200,28000,900,1500,8000]})

print("dataframe1"); display(dataframe1)
print("dataframe2"); display(dataframe2)
print("dataframe3"); display(dataframe3)
```

**Q1. Append the records of dataframe1 and dataframe3**

```python
appended = pd.concat([dataframe1, dataframe3], ignore_index=True)
print(appended.shape); appended
```

**Q2. Concatenate the data frames wherever applicable**  
(rows: df1 & df3 have the same columns → vertical; df1 & df2 → side by side, column-wise)

```python
row_concat = pd.concat([dataframe1, dataframe3], axis=0, ignore_index=True)   # same columns -> stack rows
col_concat = pd.concat([dataframe1, dataframe2], axis=1)                      # column-wise, matched by position
print("Row-wise concat:", row_concat.shape); display(row_concat.head())
print("Column-wise concat (df1 + df2):"); display(col_concat)
```

**Q3. Merge dataframe1 and dataframe2 using ProductID**

```python
merged = pd.merge(dataframe1, dataframe2, on="ProductID")
merged
```

**Q4. Join operations between dataframe1 and dataframe2**

```python
print("INNER JOIN"); display(pd.merge(dataframe1, dataframe2, on="ProductID", how="inner"))
print("LEFT JOIN");  display(pd.merge(dataframe1, dataframe2, on="ProductID", how="left"))
print("RIGHT JOIN"); display(pd.merge(dataframe1, dataframe2, on="ProductID", how="right"))
print("FULL OUTER JOIN"); display(pd.merge(dataframe1, dataframe2, on="ProductID", how="outer"))
```

**Data Transformations**  
The vector is named after your first name. *Change `Arun` to your first name and `REG_LAST4` to the last 4 digits of your Registration Number.*

```python
REG_LAST4 = 4521                      # <-- last four digits of your registration number
Arun = np.array([REG_LAST4, 12, 25, 7, 43, 18, 36, 9, 51, 30], dtype=float)   # <-- rename Arun to your first name
Fname = Arun                          # generic alias used below
print("Vector:", Fname)
```

**Q1. Normalize to [-3, +3], [-1, +1] and [0, 1] (Min-Max)**

```python
def rescale(x, new_min, new_max):
    return new_min + (x - x.min()) * (new_max - new_min) / (x.max() - x.min())

res = pd.DataFrame({"Original": Fname,
                    "[-3,+3]": rescale(Fname, -3, 3),
                    "[-1,+1]": rescale(Fname, -1, 1),
                    "[0,1] MinMax": rescale(Fname, 0, 1)})
res.round(4)
```

**Q2. Z-score transformation**

```python
z = (Fname - Fname.mean()) / Fname.std(ddof=1)      # sample std (same as R scale())
pd.DataFrame({"Original": Fname, "Z-score": z}).round(4)
```

**Q3. Decimal Scaling Normalization**  (x / 10^j, j = smallest integer with max|x'| < 1)

```python
j = len(str(int(np.abs(Fname).max())))
dec = Fname / (10 ** j)
print("j =", j)
pd.DataFrame({"Original": Fname, "Decimal Scaled": dec})
```
